```python
from keras.preprocessing import image

# Let's plot a few images
train_path = "input/train"
validation_path = "input/validation"
test_path = "input/test"

image_categories = os.listdir('input/train')

def plot_images(image_categories):
    
    # Create a figure
    plt.figure(figsize=(12, 12))
    for i, cat in enumerate(image_categories):
        
        # Load images for the ith category
        image_path = train_path + '/' + cat
        images_in_folder = os.listdir(image_path)
        first_image_of_folder = images_in_folder[0]
        first_image_path = image_path + '/' + first_image_of_folder
        img = image.load_img(first_image_path)
        img_arr = image.img_to_array(img)/255.0
        
        
        # Create Subplot and plot the images
        plt.subplot(4, 4, i+1)
        plt.imshow(img_arr)
        plt.title(cat)
        plt.axis('off')
        
    plt.show()

# Call the function
plot_images(image_categories)
```

```python
# Creating Image Data Generator for train, validation and test set
from tensorflow.keras.applications.resnet50 import preprocess_input

IMG_SIZE = (224, 224)

train_ds = tf.keras.preprocessing.image_dataset_from_directory(
    train_path,
    image_size=IMG_SIZE,
    batch_size=32,
    label_mode = 'int'
)
valid_ds = tf.keras.preprocessing.image_dataset_from_directory(
    validation_path,
    image_size=IMG_SIZE,
    batch_size=32,
    label_mode = 'int'
)
test_ds = tf.keras.preprocessing.image_dataset_from_directory(
    test_path,
    image_size=IMG_SIZE,
    batch_size=32,
    label_mode = 'int'
)
```

```python
num_classes = len(train_ds.class_names)
classNames = train_ds.class_names
print("Number of classes:", num_classes)

train_ds = train_ds.map(lambda x, y: (preprocess_input(x), y))
valid_ds = valid_ds.map(lambda x, y: (preprocess_input(x), y))
test_ds  = test_ds.map(lambda x, y: (preprocess_input(x), y))
```

Experiment with simple models and optimized datasets

```python
class VegetableCNNExperiment:
    def __init__(self, train, validate, test, class_names):
        self.train_ds = train
        self.valid_ds = validate
        self.test_ds = test
        self.class_names = class_names
        self.models = {}
        self.histories = {}
        print(f"Found {len(self.class_names)} classes: {self.class_names}")
        print(f"Training batches: {len(self.train_ds)}")
        print(f"Validation batches: {len(self.valid_ds)}")
        print(f"Test batches: {len(self.test_ds)}")

        #Optimize dataset for performance
        AUTOTUNE = tf.data.AUTOTUNE
        self.train_ds = self.train_ds.cache().prefetch(buffer_size=AUTOTUNE)
        self.valid_ds = self.valid_ds.cache().prefetch(buffer_size=AUTOTUNE)
        self.test_ds = self.test_ds.cache().prefetch(buffer_size=AUTOTUNE)
    
    def visualize_samples(self, num_samples=9):
        plt.figure(figsize=(12,12))

        for images, labels in self.train_ds.take(1):
            for i in range(min(num_samples, len(images))):
                plt.subplot(3,3,i+1)
                plt.imshow(images[i].numpy().astype("uint8"))
                class_idx = labels[i]
                plt.title(f"{self.class_names[class_idx]}")
                plt.axis('off')

        plt.tight_layout()
        plt.show()

    def create_data_augmentation(self):
        return tf.keras.Sequential([
            layers.RandomFlip("horizontal"),
            layers.RandomRotation(0.05),
            layers.RandomZoom(0.05)
        ])
    
    def create_simple_cnn(self, name = "simple_cnn"):
        model = models.Sequential([
            layers.Conv2D(32,(3,3), activation='relu'),
            layers.MaxPooling2D((2,2)),
            layers.Conv2D(64, (3,3), activation='relu'),
            layers.MaxPooling2D((2,2)),
            layers.Conv2D(64, (3,3), activation='relu'),
            layers.Flatten(),
            layers.Dense(64, activation='relu'),
            layers.Dense(len(self.class_names), activation='softmax')            
        ], name = name)

        return model
    
    def create_deeper_cnn(self, name = "deeper_cnn"):
        model = models.Sequential([
            layers.Conv2D(32,(3,3), activation='relu'),
            layers.Conv2D(32,(3,3), activation='relu'),
            layers.MaxPooling2D((2,2)),
            layers.Conv2D(64, (3,3), activation='relu'),
            layers.Conv2D(64,(3,3), activation='relu'),
            layers.MaxPooling2D((2,2)),
            layers.Conv2D(128, (3,3), activation='relu'),
            layers.Conv2D(128, (3,3), activation='relu'),
            layers.MaxPooling2D((2,2)),
            layers.Conv2D(256, (3,3), activation='relu'),
            layers.GlobalAveragePooling2D(),
            layers.Dropout(0.5),
            layers.Dense(128, activation = 'relu'),
            layers.Dropout(0.5),
            layers.Dense(len(self.class_names), activation='softmax')            
        ], name = name)

        return model
    
    def create_cnn_with_augmentation(self, name = "cnn_with_aug"):
        data_augmentation = self.create_data_augmentation()

        model = models.Sequential([
            data_augmentation,
            layers.Conv2D(32,(3,3), activation='relu'),
            layers.MaxPooling2D((2,2)),
            layers.Conv2D(64, (3,3), activation='relu'),
            layers.MaxPooling2D((2,2)),
            layers.Conv2D(128, (3,3), activation='relu'),
            layers.MaxPooling2D((2,2)),
            layers.Conv2D(128, (3,3), activation='relu'),
            layers.GlobalAveragePooling2D(),
            layers.Dropout(0.5),
            layers.Dense(128, activation = 'relu'),
            layers.Dense(len(self.class_names), activation='softmax')            
        ], name = name)

        return model
    
    def create_resnet_feature_extractor(self, name = "resnet_frozen"):
        base_model = ResNet50(
            weights = 'imagenet',
            include_top = False,
            input_shape = (224, 224, 3)
        )
        base_model.trainable = False
        model = models.Sequential([
            base_model,
            layers.GlobalAveragePooling2D(),
            layers.Dropout(0.2),
            layers.Dense(128, activation = 'relu'),
            layers.Dropout(0.2),
            layers.Dense(len(self.class_names), activation = 'softmax')
        ], name = name)
        return model
    
    def create_resnet_fine_tuned(self, name = "resnet_fine_tuned"):
        base_model = ResNet50(
            weights = 'imagenet',
            include_top = False,
            input_shape = (224, 224, 3)
        )
        base_model.trainable = True
        fine_tune_ind = 100
        for layer in base_model.layers[:fine_tune_ind]:
            layer.trainable = False
        model = models.Sequential([
            base_model,
            layers.GlobalAveragePooling2D(),
            layers.Dropout(0.2),
            layers.Dense(128, activation='relu'),
            layers.Dropout(0.2),
            layers.Dense(len(self.class_names), activation='softmax')
        ], name=name)
        return model
    
    def create_resnet_with_augmentation(self, name = "resnet_with_aug"):
        data_augmentation = self.create_data_augmentation()
        base_model = ResNet50(
            weights = 'imagenet',
            include_top = False,
            input_shape = (224, 224, 3)
        )
        base_model.trainable = True
        fine_tune_ind = 100
        for layer in base_model.layers[:fine_tune_ind]:
            layer.trainable = False
        model = models.Sequential([
            data_augmentation,
            base_model,
            layers.GlobalAveragePooling2D(),
            layers.Dropout(0.2),
            layers.Dense(128, activation='relu'),
            layers.Dropout(0.2),
            layers.Dense(len(self.class_names), activation='softmax')
        ], name=name)
        return model
    
    def train_model(self, model, epochs = 20, learning_rate = 0.001, fine_tune_lr = None):
        if fine_tune_lr is not None:
            optimizer = tf.keras.optimizers.Adam(learning_rate=fine_tune_lr)
        else:
            optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)

        model.compile(
            optimizer = optimizer,
            loss = 'sparse_categorical_crossentropy',
            metrics = ['accuracy']
        )

        for sample_images, sample_labels in self.train_ds.take(1):
            _ = model(sample_images)
            break

        print(f"\n{model.name} Architecture:")
        print(f"Total parameters: {model.count_params():,}")

        trainable_params = sum([tf.keras.backend.count_params(w) for w in model.trainable_weights])
        non_trainable_params = model.count_params() - trainable_params
        print(f"Trainable parameters: {trainable_params:,}")
        print(f"Non-trainable parameters: {non_trainable_params:,}")

        model.summary()

        is_transfer_learning = 'resnet' in model.name.lower()

        if is_transfer_learning:
            callbacks = [
                tf.keras.callbacks.EarlyStopping(
                    patience = 3,
                    restore_best_weights = True,
                    monitor = 'val_accuracy',
                    mode = 'max',
                    min_delta = 0.001
                ),
                tf.keras.callbacks.ReduceLROnPlateau(
                    factor = 0.5,
                    patience = 2,
                    monitor = 'val_accuracy',
                    mode = 'max',
                    min_delta = 0.001,
                    min_lr = 1e-7
                )
            ]
        else:
            callbacks = [
                tf.keras.callbacks.EarlyStopping(
                    patience = 5,
                    restore_best_weights = True,
                    monitor = 'val_loss'
                ),
                tf.keras.callbacks.ReduceLROnPlateau(
                    factor = 0.2,
                    patience = 3,
                    monitor = 'val_loss'
                )
            ]

        print(f"\nTraining {model.name}...")
        history = model.fit(
            self.train_ds,
            epochs=epochs,
            validation_data=self.valid_ds,
            callbacks=callbacks,
            verbose=1
        )
        return history

    def plot_training_history(self, histories, metrics = ['loss', 'accuracy']):
        fig, axes = plt.subplots(1, len(metrics), figsize = (6*len(metrics), 5))
        if len(metrics) == 1:
            axes = [axes]
        
        for i, metric in enumerate(metrics):
            for name, history in histories.items():
                axes[i].plot(history.history[metric], label=f'{name} train')
                axes[i].plot(history.history[f'val_{metric}'], label=f'{name} val')
            
            axes[i].set_title(f'Model {metric.capitalize()}')
            axes[i].set_xlabel('Epoch')
            axes[i].set_ylabel(metric.capitalize())
            axes[i].legend()
            axes[i].grid(True)
        
        plt.tight_layout()
        plt.show()

    def evaluate_models(self):
        results = {}
        for name, model in self.models.items():
            test_loss, test_acc = model.evaluate(self.test_ds, verbose=0)
            results[name] = {'loss': test_loss, 'accuracy': test_acc}
            print(f"{name:20s} | Loss: {test_loss:.4f} | Accuracy: {test_acc:.4f}")
        
        return results
    
    def create_confusion_matrix(self, model, model_name):
        y_true = []
        y_pred = []
        
        for images, labels in self.test_ds:
            predictions = model.predict(images, verbose=0)
            y_true.extend(labels.numpy())
            y_pred.extend(np.argmax(predictions, axis=1))
        
        # Create confusion matrix
        cm = confusion_matrix(y_true, y_pred)
        
        plt.figure(figsize=(12, 10))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=self.class_names,
                   yticklabels=self.class_names)
        plt.title(f'Confusion Matrix - {model_name}')
        plt.xlabel('Predicted')
        plt.ylabel('Actual')
        plt.xticks(rotation=45)
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show()
        
        # Print classification report
        print(f"\nClassification Report - {model_name}:")
        print(classification_report(y_true, y_pred, target_names=self.class_names))
    
    def run_baseline_experiments(self, epochs = 20):
        models_to_train = [
            ("Simple CNN", self.create_simple_cnn(), 0.001, None),
            ("Deeper CNN", self.create_deeper_cnn(), 0.001, None),
            ("CNN with Augmentation", self.create_cnn_with_augmentation(), 0.0005, None)
        ]
        for name, model, lr, fine_tune_lr in models_to_train:
            history = self.train_model(model, epochs=epochs, learning_rate=lr, fine_tune_lr=fine_tune_lr)
            self.models[name] = model
            self.histories[name] = history
        
        return self.models, self.histories
    
    def run_transfer_learning_experiments(self, epochs=10):
        transfer_models = [
            ("ResNet50 Frozen", self.create_resnet_feature_extractor(), 0.001, None),
            ("ResNet50 Fine-tuned", self.create_resnet_fine_tuned(), 0.001, 0.0001),
            ("ResNet50 + Augmentation", self.create_resnet_with_augmentation(), 0.001, 0.0001)
        ]
        
        for name, model, lr, fine_tune_lr in transfer_models:
            history = self.train_model(model, epochs=epochs, learning_rate=lr, fine_tune_lr=fine_tune_lr)
            self.models[name] = model
            self.histories[name] = history
        
        return self.models, self.histories
    
    def run_complete_experiments(self, baseline_epochs=20, transfer_epochs=10):
        self.run_baseline_experiments(epochs=baseline_epochs)
        self.run_transfer_learning_experiments(epochs=transfer_epochs)
        # Plot comparison of all models
        self.plot_training_history(self.histories)
        # Evaluate all models
        results = self.evaluate_models()
        # Show detailed analysis
        self.analyze_results(results)
        
        return results
    
    def analyze_results(self, results):
        # Sort results by accuracy
        sorted_results = sorted(results.items(), key=lambda x: x[1]['accuracy'], reverse=True)
        
        for name, metrics in sorted_results:
            model = self.models[name]
            param_count = f"{model.count_params():,}"
            model_type = "Transfer Learning" if "ResNet" in name else "From Scratch"
            print(f"{name:<25} {metrics['accuracy']:<10.4f} {param_count:<12} {model_type:<15}")
        
        # Performance insights
        best_overall = sorted_results[0]
        best_baseline = max([(k, v) for k, v in results.items() if "ResNet" not in k], 
                          key=lambda x: x[1]['accuracy'])
        best_transfer = max([(k, v) for k, v in results.items() if "ResNet" in k], 
                          key=lambda x: x[1]['accuracy'])
        
        print(f"🏆 Best Overall: {best_overall[0]} ({best_overall[1]['accuracy']:.4f})")
        print(f"📊 Best Baseline: {best_baseline[0]} ({best_baseline[1]['accuracy']:.4f})")
        print(f"🚀 Best Transfer: {best_transfer[0]} ({best_transfer[1]['accuracy']:.4f})")
        
        # Calculate improvement
        improvement = (best_transfer[1]['accuracy'] - best_baseline[1]['accuracy']) * 100
        print(f"📈 Transfer Learning Improvement: {improvement:.2f} percentage points")
        
        # Show confusion matrix for best model
        print(f"\nConfusion Matrix for Best Model: {best_overall[0]}")
        self.create_confusion_matrix(self.models[best_overall[0]], best_overall[0])
    
```

Somethings I had to experiment with:
-Learning rate: CNN with Augmented data needed a smaller learning rate to learn -> higher learning rates were preventing the model from learning
    ->too high = model can't converge
    ->too low = learning takes forever
    ->ReduceLROnPlateau: automatically finds the right balance (auto updates the learning rate during training)
-Augmentation: had to lower the augmentation, too aggressive of augmentation also increases difficulty of model learning/converging 
-Optimized the datasets to improve performance time of training
-Used callbacks to also improve performance time of training -> stops training if training has already found peak weights or rather stops improving during epochs
-Compared layer architecture between simple and deeper layer architectures and how they affect accuracy, precision, recall, and f1-scores
    -> simple: fast to train, prone to overfitting
    -> deeper: greatly improves accuracy but it prone to overfitting quickly if data is not regularized 
        ->Dropout: regularization technique -> prevents overfitting by randomly zeroing nuerons during training
        ->EarlyStopping: automatic regularization that timeout training
    -> Augmentation: regularizes the data which helps prevent overfitting, more robust
        ->need hyperparameter tuning = playing with learning rates and augmentation techniques
        ->conservative approach works, want to preserve data integrity and not be too aggressive as to harm the learning
-Learned to balance model capacity with dataset size
-Used generalization gap to reveal overfitting
